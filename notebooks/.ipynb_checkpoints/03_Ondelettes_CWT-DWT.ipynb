# Transformée en ondelettes (CWT / DWT)

1. On peut utiliser la transformée en ondelettes discrète (DWT) pour faire l'extraction de caractéristiques (features) avec la bibliothèque pywt.

2. On peut utiliser la transformée en ondelettes continue (CWT) pour faire une analyse fine temps-fréquence (ex. visualisation, spectrogrammes, CNN) avec la bibliothèque scipy.signal.cwt.

## Fourier vs Ondelettes
Contrairement à la FFT, l'ondelette donne des informations temps-fréquence. Avec la FFT, on peut voir l’intensité des fréquences globales, ce qui est utile, mais on ne peut pas détecter les changements de fréquence dans le temps, ce qui pourrait permettre d'extraire un pattern de pensée (ex. : droite) en machine learning.

_"Pour comparer à la musique, la FFT c’est comme avoir la liste des instruments, mais sans savoir à quel moment chacun joue.
L'ondelette elle dit: Entre 1.5 et 2 secondes, y’avait un motif aigu (genre cymbales), puis un grave à 3 secondes (genre kick drum)."_
